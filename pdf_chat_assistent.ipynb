```python
from langchain_community.document_loaders import PyPDFLoader

pdf_path = "/content/Deep Learning with Python.pdf"

loader = PyPDFLoader(pdf_path)
docs = loader.load()
```

```python
! pip install langchain-text-splitters
```

```python
# chinging the docs

from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size = 500 ,
    chunk_overlap = 50
)
chunks = splitter.create_documents([full_docs])
```

```python
chunks
```

```python
!pip install langchain-huggingface sentence-transformers
```

```python
# generating embeddings
from langchain_huggingface import  HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(model_name = "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2")
```

```python
!pip install faiss-cpu
```

```python
type(chunks)
```

```python
# creating the vectore store
from langchain_community.vectorstores import FAISS

vector_store = FAISS.from_documents( chunks , embeddings)
```

```python
# similarity search

query = "Why deep learning? Why now?"

similarity_search = vector_store.similarity_search(query, k = 1)
```

```python
similarity_search
```

```python
results = vector_store.similarity_search(query, k=3)

for doc in results:
    print(doc.page_content[:500])
    print("-" * 80)
```

```python
# retriever multi query retriever

from langchain_classic.retrievers.multi_query import MultiQueryRetriever

retriever = vector_store.as_retriever()

advanced_retriever = MultiQueryRetriever.from_llm(
    retriever = retriever , llm=llm
)
```

```python
advanced_retriever
```

```python
type(advanced_retriever)
```

```python
question = "what is AI and how AI is connect to Deep learning"
```

```python
retrieved_docs = advanced_retriever.invoke(question)
```

```python
context = "\n\n".join(d.page_content for d in retrieved_docs)
```

```python
context
```

```python
# froming the prompt
from langchain_core.prompts import PromptTemplate , ChatPromptTemplate

simple_prompt_template = PromptTemplate(
    input_variables=["question", "context"],
    template="""You are an AI teacher and assistant.

ROLE:
Explain concepts clearly to a Generative AI Engineer. Assume the learner knows
Python and basic ML, and go deeper into the technical details (architecture,
trade-offs, code examples where useful).

RULES:
1. Answer ONLY using the context provided below.
2. If the context does not contain enough information to answer, reply exactly:
   "This is not in my knowledge based on the provided context."
   Do not guess or make up information.
3. Support every claim with proof: quote or cite the relevant part of the
   context under a "Proof" section.
4. Keep the explanation structured and easy to follow.

CONTEXT:
{context}

QUESTION:
{question}

ANSWER FORMAT:
- Explanation: (clear, technical, step-by-step)
- Example: (short code or practical example, only if supported by the context)
- Proof: (exact lines or snippets from the context that support your answer)
""")

```

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt_template = ChatPromptTemplate.from_messages(
    [
          ("system","""
      You are an AI teacher and assistant explaining to a Generative AI Engineer.

RULES:
1. Answer ONLY from the context below.
2. If the answer is not in the context, say:
   "This is not in my knowledge based on the provided context."
3. Do not guess or hallucinate.
4. Add a "Proof" section quoting the exact lines from the context.

CONTEXT:
{context}
  """),
(
 "human" , "{question}"
)
    ]
)
```

```python
chat_full_prompt = chat_prompt_template.invoke({"question":question , "context":context})
```

```python
full_prompt = simple_prompt_template.invoke({"question":question , "context":context})
```

```python
type(chat_full_prompt)
```

```python
chat_prompt_result = llm.invoke(chat_full_prompt)
```
